# Section 11: Integrated Lab Questions

```python
import numpy as np

# Q39. Student Marks Analysis: Create a NumPy array containing marks of 10 students in 3 subjects.
# Calculate each student's total and average marks. Identify students whose average is >= 50 using np.where().
np.random.seed(5)
marks_matrix = np.random.randint(30, 100, size=(10, 3))
total_marks = np.sum(marks_matrix, axis=1)
average_marks = np.mean(marks_matrix, axis=1)
pass_students = np.where(average_marks >= 50)[0]

print("Q39:")
print("Marks (10 Students x 3 Subjects):\n", marks_matrix)
print("Total Marks:", total_marks)
print("Average Marks:", np.round(average_marks, 2))
print("Indices of students with Average >= 50:", pass_students)
print("-" * 40)
```

```python
import numpy as np

# Q40. Matrix Analysis: Create a 4 x 4 matrix containing numbers 1-16.
# Find its transpose, row-wise sums, column-wise sums, mean, maximum, and minimum. Extract the central 2 x 2 matrix.
mat_4x4 = np.arange(1, 17).reshape(4, 4)
central_mat = mat_4x4[1:3, 1:3]

print("Q40:")
print("4x4 Matrix:\n", mat_4x4)
print("Transpose:\n", mat_4x4.T)
print("Row-wise Sums:", np.sum(mat_4x4, axis=1))
print("Column-wise Sums:", np.sum(mat_4x4, axis=0))
print("Overall Mean:", np.mean(mat_4x4))
print("Overall Max:", np.max(mat_4x4))
print("Overall Min:", np.min(mat_4x4))
print("Central 2x2 Matrix:\n", central_mat)
print("-" * 40)
```

```python
import numpy as np

# Calculate statistics for the central 2x2 matrix from Q40
central_sum = np.sum(central_mat)
central_mean = np.mean(central_mat)
central_max = np.max(central_mat)
central_min = np.min(central_mat)
central_std = np.std(central_mat)

print("Statistics for the Central 2x2 Matrix:")
print("Central Matrix:\n", central_mat)
print("Sum:", central_sum)
print("Mean:", central_mean)
print("Max:", central_max)
print("Min:", central_min)
print("Standard Deviation:", np.round(central_std, 4))
```

```python
import numpy as np

# Transpose the central 2x2 matrix from Q40
transposed_central = central_mat.T
mean_transposed_central = np.mean(transposed_central)

print("Transposed Central Matrix:\n", transposed_central)
print("Mean of Transposed Central Matrix:", mean_transposed_central)
```

```python
import numpy as np

# Q41. Random Data Analysis: Generate 20 observations using np.random.randn().
# Calculate mean, standard deviation, minimum, maximum, and sorted values.
# Replace all negative observations with zero using np.where().
np.random.seed(8)
obs = np.random.randn(20)
sorted_obs = np.sort(obs)
positive_obs = np.where(obs < 0, 0, obs)

print("Q41:")
print("Generated Observations:\n", obs)
print("Mean:", np.mean(obs))
print("Standard Deviation:", np.std(obs))
print("Minimum:", np.min(obs))
print("Maximum:", np.max(obs))
print("Sorted Observations:\n", sorted_obs)
print("Negative Values Replaced with Zero:\n", positive_obs)
print("-" * 40)
```

```python
import numpy as np

# Q42. Matrix Operations Challenge: Create two 3 x 3 matrices and perform:
# Element-wise addition, element-wise multiplication, matrix multiplication,
# transpose of the first matrix, and maximum/minimum of the resulting matrix.
np.random.seed(12)
M1 = np.random.randint(1, 10, size=(3, 3))
M2 = np.random.randint(1, 10, size=(3, 3))
matmul_res = np.dot(M1, M2)

print("Q42:")
print("Matrix M1:\n", M1)
print("Matrix M2:\n", M2)
print("Element-wise Addition:\n", M1 + M2)
print("Element-wise Multiplication:\n", M1 * M2)
print("Matrix Multiplication:\n", matmul_res)
print("Transpose of M1:\n", M1.T)
print("Max of resulting product matrix:", np.max(matmul_res))
print("Min of resulting product matrix:", np.min(matmul_res))
print("-" * 40)
```

```python
import numpy as np

# Q43. Array Manipulation Challenge: Create an array containing 1-24, reshape it into a 4 x 6 matrix,
# extract selected rows and columns, transpose it, calculate row/column statistics, and identify unique values.
arr_24 = np.arange(1, 25).reshape(4, 6)
extracted_part = arr_24[0:2, 1:4] # rows 0-1, cols 1-3
transposed_part = extracted_part.T

print("Q43:")
print("Reshaped 4x6 Matrix:\n", arr_24)
print("Extracted rows 0-1 and cols 1-3:\n", extracted_part)
print("Transposed Part:\n", transposed_part)
print("Row-wise sum of reshaped matrix:", np.sum(arr_24, axis=1))
print("Column-wise mean of reshaped matrix:", np.mean(arr_24, axis=0))
print("Unique Values in the matrix:", np.unique(arr_24))
```

```python
import numpy as np

# Q12. Create an array and demonstrate the difference between a slice/view and a copy.
original = np.array([1, 2, 3, 4, 5])

# View (Slice)
view_arr = original[1:4]
view_arr[0] = 99

# Copy
copy_arr = original.copy()
copy_arr[1] = 500

print("Q12:")
print("Original (Note index 1 changed because of the view modification):", original)
print("View modified:", view_arr)
print("Copy modified (Original remains unaffected by this):", copy_arr)
print("-" * 40)
```
