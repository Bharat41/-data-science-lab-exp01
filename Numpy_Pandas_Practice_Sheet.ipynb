```python
import numpy as np

arr = np.arange(11, 30, 2)
print("Array:", arr)
print("Number of dimensions:", arr.ndim)
print("Total number of elements:", arr.size)
```

```python
vals = np.linspace(5, 35, 7)
print("Values:", vals)
```

```python
arr = np.full((4, 3), 9)
# Third row is index 2, second column is index 1
arr[2, 1] = 0
print(arr)
```

```python
a = np.array([6, 12, 18, 24])
remainder = a % 5  # or np.remainder(a, 5)
print("Remainder:", remainder)
```

```python
arr = np.array([14, 28, 42, 56, 70, 84])
# Indices 1, 3, 5 with step 2
subset = arr[1::2]
print("Extracted elements:", subset)
```

```python
arr = np.array([-16, 25, 36, 49])
abs_arr = np.abs(arr)
sqrt_arr = np.sqrt(abs_arr)
print("Absolute values:", abs_arr)
print("Square roots:", sqrt_arr)
```

```python
mat = np.arange(7, 13).reshape(3, 2)
print("Original shape:", mat.shape)
trans = mat.T
print("Transposed matrix:\n", trans)
print("Shape after transpose:", trans.shape)
```

```python
temperatures = np.array([18, 22, 27, 31, 24, 19])
temp_range = temperatures.max() - temperatures.min()  # or np.ptp(temperatures)
median_temp = np.median(temperatures)
print("Range:", temp_range)
print("Median:", median_temp)
```

```python
speed = np.array([35, 62, 48, 81, 55])
category = np.where(speed >= 60, 'Fast', 'Normal')
print("Categories:", category)
```

```python
values = np.array([4, 9, 4, 2, 9, 7, 2, 4])
distinct_vals = np.unique(values)
print("Distinct values (sorted):", distinct_vals)
print("Count of distinct values:", len(distinct_vals))
```

```python
import pandas as pd

days = ['Monday', 'Tuesday', 'Wednesday', 'Thursday']
books = pd.Series([24, 31, 19, 27], index=days)
print(books)
```

```python
prices = pd.Series([45000, 18000, 22000], index=['Laptop', 'Tablet', 'Phone'])
tablet_price = prices['Tablet']
print("Tablet price:", tablet_price)
```
