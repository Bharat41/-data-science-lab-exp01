```python
rng = np.random.default_rng(42)
marks = rng.integers(20, 101, size=(12, 4))
student_avg = marks.mean(axis=1)
test_mean = marks.mean(axis=0)
test_std = marks.std(axis=0)
passing = np.sum(np.all(marks >= 40, axis=1))
high_avg_students = np.where(student_avg >= 75)[0] + 1
max_variation_test = np.argmax(test_std) + 1
print("Averages:", student_avg)
print("Test Mean:", test_mean)
print("Test Std:", test_std)
print("Passing all:", passing)
print("High Averages Students:", high_avg_students)
print("Max variation test:", max_variation_test)
```

```python
matrix = np.arange(21, 46).reshape(5, 5)
central = matrix[1:4, 1:4]
central_t = central.T
prod = np.dot(central, central_t)
row_sums = prod.sum(axis=1)
max_sum = np.max(row_sums)
tied_pos = np.argwhere(row_sums == max_sum).flatten()
print(matrix.shape)
print(central.shape)
print(central_t.shape)
print(prod.shape)
print(row_sums)
print(max_sum)
print(tied_pos)
```

```python
quantities = np.array([
    [10, 15, 20, 8],
    [12, 22, 18, 15],
    [5, 10, 15, 30],
    [14, 25, 20, 12],
    [30, 45, 10, 5],
    [8, 12, 40, 50]
])
prices = np.array([150, 240, 90, 320])
daily_rev = np.dot(quantities, prices)
prod_rev = quantities.sum(axis=0) * prices
best_day = np.argmax(daily_rev) + 1
best_prod = np.argmax(quantities.sum(axis=0)) + 1
total_qty = quantities.sum(axis=0)
new_prices = np.where(total_qty > 100, prices * 0.9, prices)
new_total_rev = np.sum(quantities * new_prices)
print(daily_rev)
print(prod_rev)
print(best_day)
print(best_prod)
print(new_total_rev)
```

```python
rng = np.random.default_rng(42)
obs = rng.normal(50, 12, 100)
outside_count = np.sum((obs < 30) | (obs > 70))
cleaned = np.where(obs < 30, 30, np.where(obs > 70, 70, obs))
print(outside_count)
for arr in [obs, cleaned]:
    print(np.mean(arr), np.var(arr), np.std(arr), np.min(arr), np.max(arr))
```

```python
branch_a = pd.Series({'Q1': 120, 'Q2': 145, 'Q4': 180})
branch_b = pd.Series({'Q2': 110, 'Q3': 135, 'Q4': 155})
branch_c = pd.Series({'Q1': 95, 'Q3': 125, 'Q4': 140})
print(branch_a + branch_b + branch_c)
common_idx = branch_a.index.intersection(branch_b.index).intersection(branch_c.index)
print(common_idx)
print((branch_a + branch_b + branch_c)[common_idx])
filled_sum = branch_a.add(branch_b, fill_value=0).add(branch_c, fill_value=0)
print(filled_sum)
```
