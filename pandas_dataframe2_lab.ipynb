# Pandas DataFrame Lab - Solutions
Sections A (Easy), B (Moderate), C (Hard)

```python
import pandas as pd
import numpy as np
from io import StringIO

def clip_or_sample(sample_text):
    # Reads from clipboard; falls back to sample text if clipboard is unavailable
    try:
        return pd.read_clipboard()
    except Exception:
        return pd.read_csv(StringIO(sample_text), sep=r"\s{2,}|\t|,", engine="python")
```

**26. `course_df`: columns, Student, Student+Score, add Batch.**

**27. `travel_df`: Code via `np.arange(5)`, then replace with Series at indexes 1 and 4; display after each.**

**28. `school_df`; request Student, Class, Section, House; assign one common House value.**

```python
school_df = pd.DataFrame({
    'Student': ['A', 'B', 'C', 'D'],
    'Class': [10, 10, 11, 12],
    'Section': ['A', 'B', 'A', 'C']
})
school_house_df = pd.DataFrame(school_df, columns=['Student', 'Class', 'Section', 'House'])
print(school_house_df)   # House is NaN
school_house_df['House'] = 'Blue'
school_house_df
```

**29. `inventory_df`: partially indexed Series column; display; delete; display again.**

```python
inventory_df = pd.DataFrame({
    'Item': ['Laptop', 'Mouse', 'Keyboard', 'Monitor'],
    'Quantity': [10, 50, 30, 15],
    'Price': [55000, 500, 1200, 9000]
})
inventory_df['Discount'] = pd.Series([10, 5], index=[0, 2])
print(inventory_df)
del inventory_df['Discount']
inventory_df
```

**30. `lab_df` from a copied 4+ column table; full sequence of operations.**

```python
lab_df = clip_or_sample('''Name,Dept,Total Marks,City
Aman,IT,450,Delhi
Riya,HR,480,Pune
Karan,Sales,390,Mumbai''')
print(lab_df)                                                    # display
print(lab_df.columns)                                            # column names
print(lab_df['Name'])                                            # single-word column
print(lab_df['Total Marks'])                                     # column with a space
small_df = pd.DataFrame(lab_df, columns=['Name', 'Total Marks']) # smaller DataFrame
print(small_df)
lab_df['Status'] = 'Active'                                      # add whole column
print(lab_df)
del lab_df['Status']                                             # delete column
lab_df
```
