```python
counter = 0
def show_counter():
    print(counter)

show_counter()
```

```python
counter = 0
def increment_counter():
    counter = counter + 1

try:
    increment_counter()
except UnboundLocalError as e:
    print("UnboundLocalError:", e)
```

```python
counter = 0
def increment_counter():
    global counter
    counter = counter + 1

increment_counter()
increment_counter()
print(counter)   # 2
```
