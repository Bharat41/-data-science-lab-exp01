## Section B – Medium (10 Questions)

```python
# Q16: Predict which labelled entries will be returned by the following expression.
s = Series([18, 45, 27, 61], index=['P','Q','R','S'])
# Prediction: Q (45) and S (61)
print("Q16 Result:")
print(s[s > 30])
```

```python
# Q18: Create a Series using ordered index Delhi, Kochi, Pune, Chennai. State which label will receive NaN.
city_data = {'Delhi': 80, 'Pune': 65, 'Kochi': 70}
city_series = Series(city_data, index=['Delhi', 'Kochi', 'Pune', 'Chennai'])
print("Q18 (Chennai receives NaN because it is missing in city_data):")
display(city_series)
```

```python
# Q19: State the Boolean output produced by pd.isnull(s)
s = Series({'W': 5, 'X': 10, 'Y': 15}, index=['W','X','Y','Z'])
# Output conceptually: W: False, X: False, Y: False, Z: True
print("Q19:")
print(pd.isnull(s))
```

```python
# Q20: State the Boolean output produced by pd.notnull(s)
# Output conceptually: W: True, X: True, Y: True, Z: False
print("Q20:")
print(pd.notnull(s))
```

```python
# Q21: Predict the result of adding the following two Series.
s1 = Series([10, 20, 30], index=['A','B','C'])
s2 = Series([1, 2, 3], index=['C','A','B'])
# Alignment matching:
# A: 10 + 2 = 12
# B: 20 + 3 = 23
# C: 30 + 1 = 31
print("Q21:")
print(s1 + s2)
```

```python
# Q22: Explain why addition produces missing values.
s1 = Series({'A': 12, 'B': 18})
s2 = Series({'B': 5, 'C': 9})
# Explanation: 'A' is only in s1, and 'C' is only in s2.
# Since addition requires matching keys in both Series, unmatched keys result in NaN.
print("Q22:")
print(s1 + s2)
```

## Section C – Hard (5 Questions)

```python
# Q26: Analyse index alignment.
north = Series({'Jan': 40, 'Feb': 55, 'Mar': 60})
south = Series({'Feb': 20, 'Mar': 25, 'Apr': 30})
total = north + south
# Explanation:
# - 'Jan' and 'Apr' exist in only one Series, so they align with NaN and output NaN.
# - 'Feb' aligns 55 + 20 = 75.0
# - 'Mar' aligns 60 + 25 = 85.0
print("Q26:")
display(total)
```

```python
# Q27: Study the code and answer all parts
d = {'Red': 12, 'Blue': 18, 'Green': 24}
labels = ['Green','Red','Yellow','Blue']
result = Series(d, index=labels)

print("Q27 Outputs:")
print(f"(a) Label containing NaN: Yellow")
print(f"(b) pd.isnull(result) for Yellow: {pd.isnull(result)['Yellow']}")
print("(c) Reason: 'Yellow' is not present in the original dictionary keys, so Pandas pads it with NaN.")
print("(d) Labels retaining numeric values: Green, Red, Blue")
```

```python
# Q28: Trace transformation
s = Series([90, 75, 82], index=['Ravi','Aman','Neha'])
d = s.to_dict()
s2 = Series(d)

# Trace: s is converted to dict {'Ravi': 90, 'Aman': 75, 'Neha': 82}.
# s2 is created from dict, which maintains the same keys and values.
# Do they represent the same labelled data? Yes.
print("Q28:")
display(s2)
```

```python
# Q29: Write a short Pandas program fulfilling multiple tasks.
# Create a labelled Series
products = Series([450, 600, 300, 750], index=['A', 'B', 'C', 'D'])
print("Original:", products.values, products.index)
# Select one label
print("Value of B:", products['B'])
# Filter above 500
print("Filtered:", products[products > 500])
# Check existence
print("Is 'A' in Series?", 'A' in products)
# Convert to dictionary
print("As Dictionary:", products.to_dict())
```

```python
# Q30: Analyse code and answer parts (a) to (e)
data = {'A': 15, 'B': 35, 'C': 25}
s = Series(data, index=['C','A','D','B'])
s.name = 'Observations'
s.index.name = 'Category'

print("Q30 Outputs:")
print("(a) Index order:", list(s.index))
print("(b) Label containing NaN: D")
print("(c) Boolean filter s[s > 20]:\n", s[s > 20])
print("(d) pd.notnull(s) indicates:\n", pd.notnull(s))
print(f"(e) Series name: {s.name}, Index name: {s.index.name}")
```
