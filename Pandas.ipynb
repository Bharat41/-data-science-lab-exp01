```python
products = Series([450, 600, 300, 750], index=['A', 'B', 'C', 'D'])
print("Original:", products.values, products.index)
print("Value of B:", products['B'])
print("Filtered:", products[products > 500])
print("Is 'A' in Series?", 'A' in products)
print("As Dictionary:", products.to_dict())
```

```python
data = {'A': 15, 'B': 35, 'C': 25}
s = Series(data, index=['C','A','D','B'])
s.name = 'Observations'
s.index.name = 'Category'
print("(a) Index order:", list(s.index))
print("(b) Label containing NaN: D")
print("(c) Boolean filter s[s > 20]:\n", s[s > 20])
print("(d) pd.notnull(s) indicates:\n", pd.notnull(s))
print(f"(e) Series name: {s.name}, Index name: {s.index.name}")
```
